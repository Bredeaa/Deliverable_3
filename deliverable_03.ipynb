```python
# Import classes from Python files

from characters import Character
from dice import Dice
from items import Weapon
from warrior import Warrior
from mage import Mage
from battle import Battle
```

```python
# Task 3: Demonstrate Weapon class

# Create two weapons
sword = Weapon("Sword", Dice(20))
staff = Weapon("Staff", Dice(16))

# Show weapon attributes
print("Sword name:", sword.name)
print("Sword dice:", sword.damage_dice)

print("Staff name:", staff.name)
print("Staff dice:", staff.damage_dice)
```

```python
# Task 4: Demonstrate Warrior class

# Create a warrior
warrior = Warrior("Arthur")

# Show attributes
print("Name:", warrior.name)
print("HP:", warrior.hp)
print("Base damage:", warrior.damage)
print("Armor:", warrior.armor)


# Create a mage 
mage = Mage("Merlin")

# Show attributes
print("\nName:", mage.name)
print("HP:", mage.hp)
print("Base damage:", mage.damage)
print("Mana:", mage.mana)
print("Maximum mana:", mage.mana_max)

```

```python
# Task 5: Simulate battles

# Battle with weapons
print("BATTLE WITH WEAPONS")

hero = Warrior("Arthur", weapon=sword)
enemy = Mage("Merlin", weapon=staff)

battle = Battle(hero, enemy)
battle.simulate_battle()


# Battle without weapons
print("\nBATTLE WITHOUT WEAPONS")

hero2 = Warrior("Arthur")
enemy2 = Mage("Merlin")

battle2 = Battle(hero2, enemy2)
battle2.simulate_battle()
```
